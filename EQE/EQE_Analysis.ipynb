# EQE Analysis and Jsc Calculation

This notebook analyzes the External Quantum Efficiency (EQE) data and calculates the short-circuit current density (Jsc) using the AM1.5G solar spectrum.

## Formula
$$J_{sc} = \frac{q}{hc} \int_{\lambda_1}^{\lambda_2} EQE(\lambda) \times \lambda \times \Phi_{AM1.5G}(\lambda) \, d\lambda$$

where:
- **EQE(λ)**: External Quantum Efficiency
- **Φ_AM1.5G(λ)**: AM1.5G solar spectrum irradiance  
- **λ**: Wavelength
- **Integration range**: 300 - 1100 nm

This program takes the data from the EQE spectra, multiply it for the respective AM1.5G spectra and take the numerical integral by applying the simpson's rule

```python
def interpolate_values(x, y, new_x):
    """
    Interpolate y values at new x positions using linear interpolation.
    """
    f = interpolate.interp1d(x, y, kind='linear', fill_value='extrapolate')
    return f(new_x)

def calculate_jsc(eqe_data, am15g_spectrum):
    """
    Calculate short-circuit current density (Jsc) from EQE data and AM1.5G spectrum.
    
    Parameters:
    -----------
    eqe_data : DataFrame
        Must have columns 'WL' (wavelength in nm) and 'QE' (quantum efficiency)
    am15g_spectrum : DataFrame
        Must have columns 'WL' (wavelength in nm) and 'AM1.5G' (irradiance in W/m²/nm)
    
    Returns:
    --------
    Jsc : float
        Short-circuit current density in mA/cm²
    """
    df_eqe = eqe_data.copy()
    df_am15 = am15g_spectrum.copy()
    
    # Convert QE to fraction if it's in percentage
    if df_eqe['QE'].mean() > 1:
        df_eqe['QE'] = df_eqe['QE'] / 100.0
    
    # Set index and crop AM1.5G spectrum to EQE wavelength range
    df_eqe.set_index('WL', inplace=True)
    df_am15.set_index('WL', inplace=True)
    df_am15 = df_am15[(df_am15.index >= df_eqe.index.min()) & (df_am15.index <= df_eqe.index.max())]
    
    # Interpolate EQE to match AM1.5G wavelength grid
    f_eqe = interpolate.interp1d(df_eqe.index, df_eqe['QE'], kind='linear', fill_value='extrapolate')
    df_am15['EQE_interp'] = f_eqe(df_am15.index)
    
    # Calculate integrand: EQE(λ) × AM1.5G(λ) × λ / (hc)
    # hc = 1239.84193 eV·nm
    df_am15['integrand'] = df_am15['AM1.5G'] * df_am15.index / 1239.84193 * df_am15['EQE_interp']
    
    # Integrate using Simpson's rule and convert to mA/cm²
    jsc = integrate.simpson(df_am15['integrand'], x=df_am15.index) / 10.0
    
    return jsc

print("Helper functions defined successfully!")
```

### Parameter Dependencies on Jsc Correction Factor

The following parameters are corrected based on their relationship to the Jsc value determined from the EQE measurement:

#### **1. Short-Circuit Current Density (Jsc)**
$$J_{sc,corrected} = J_{sc,original} \times CF$$

- **Direction**: Multiplied by correction factor
- **Reason**: Directly proportional to the EQE-derived Jsc. If the correction factor > 1, the original Jsc was underestimated.

#### **2. Fill Factor (FF)**
$$FF_{corrected} = {FF_{original}}$$

- **Direction**: Divided by correction factor
- **Reason**: The Fill Factor represents the "squareness" of the J-V curve, calculated as $P_{max} / (V_{oc} \times J_{sc})$. Since applying a correction factor scales both the maximum power ($P_{max}$) and the short-circuit current ($J_{sc}$) by the exact same proportion, this factor cancels out. The intrinsic shape of the curve is preserved, keeping the FF constant.

#### **3. Power Conversion Efficiency (PCE/η)**
$$\eta_{corrected} = \eta_{original} \times CF$$

- **Direction**: Multiplied by correction factor
- **Reason**: Direct product of Jsc (among other factors). Since PCE ∝ Jsc, the efficiency scales proportionally with the Jsc correction factor.

#### **4. Open-Circuit Voltage (Voc)**
$$V_{oc,corrected} = V_{oc,original}$$

No correction applied.

- **Direction**: Unchanged
- **Reason**: Voc depends primarily on the bandgap and recombination properties of the material, not on the absolute photocurrent value. It remains constant regardless of Jsc scaling.

---

**Note**: The correction factor is calculated as:

$$CF = \frac{J_{sc,EQE}}{J_{sc,solar simulator}}$$

# Band-Gap Extraction

The optical band gap is determined from the EQE spectrum using a Tauc-like linear analysis. First, the wavelength data are converted to photon energy using

$$
E = \frac{hc}{\lambda}
$$

where $E$ is in eV and $\lambda$ is in nm. 

Then, the band gap can be obtained by doing a linear fit and getting the intercept of this line with the x-axis:

$$
(\ln{(1-EQE) \cdot E})^2 \propto E - E_g
$$

which is plotted as a function of photon energy. In the absorption edge region, the data follow an approximately linear trend, and the optical band gap is obtained from the x-intercept of the straight-line fit, i.e., the energy at which the fitted line crosses the horizontal axis.

In practice, the code identifies the inflection point of the Tauc curve by combining:
- Savitzky–Golay smoothing of the signal,
- first and second derivatives to locate the region of maximum slope and the zero-crossing of curvature,
- a linear extrapolation of the near-edge region to obtain the intercept energy.

The extracted value is therefore the extrapolated transition energy corresponding to the onset of strong absorption.

This approach is based on the work of [Keller, J.](https://www.researchgate.net/publication/352019926_Fundamental_characterization_of_thin-film_solar_cells_IV_and_QE), that is a standard Tauc analysis method commonly used for optical band-gap determination in semiconductor materials. 

## Uncertainty Calculation

The uncertainty in the band-gap value is estimated from the linear fit used in the extrapolation. The code performs a linear regression in a narrow energy window around the selected absorption edge and uses the standard error of the regression to propagate the uncertainty in the fitted slope and intercept. Since the band gap is calculated as the x-intercept of the line,

$$
E_g = -\frac{c}{m}
$$

where $m$ is the slope and $c$ is the intercept, the uncertainty is obtained by standard error propagation:

$$
\sigma_{E_g} = \sqrt{\left(\frac{\partial E_g}{\partial c}\sigma_c\right)^2 + \left(\frac{\partial E_g}{\partial m}\sigma_m\right)^2}
$$

with

$$
\frac{\partial E_g}{\partial c} = -\frac{1}{m},
\qquad
\frac{\partial E_g}{\partial m} = \frac{c}{m^2}
$$

so the reported value is written as:

$$
E_g \pm \sigma_{E_g}
$$

This gives the band-gap value together with its uncertainty from the linear fitting procedure.


```python
def linear_fitting(x, y, title, df=None):
    """Performs Tauc plot linear fitting by finding the maximum gradient with error analysis."""
    fig, ax = plt.subplots(figsize=(8, 6))

    # Sort data by x (Energy) to ensure interp1d works correctly 
    # (Since E = 1240/lambda, E decreases as lambda increases)
    sort_idx = np.argsort(x)
    x_tot = x[sort_idx]
    y_tot = y[sort_idx]

    tauc_spectrum = np.column_stack((x_tot, y_tot))

    # Smooth and interpolate
    y_func = interp1d(tauc_spectrum[:, 0], savgol_filter(tauc_spectrum[:, 1], 11, 3))
    x_smooth = np.linspace(tauc_spectrum[0, 0], tauc_spectrum[-1, 0], 5000)

    # Calculate 1st derivative along Tauc plot
    dy = np.diff(y_func(x_smooth), 1, 0)
    dx = np.diff(x_smooth, 1, 0)
    y_1d = interp1d(x_smooth[:-1], dy / dx)

    # Calculate 2nd derivative along Tauc plot
    d2y = np.diff(y_func(x_smooth), 2, 0)
    dx2 = x_smooth[1] - x_smooth[0]
    y_2d = interp1d(x_smooth[:-2], d2y / dx2)

    # Find point where 2nd derivative == 0 and gradient is at a maximum
    gradmax = 0.
    x_0 = x_smooth[len(x_smooth)//2] # Fallback value
    y_0 = y_func(x_0)

    for i in range(2, len(x_smooth[:-2])):
        grad = y_1d(x_smooth[:-2])[i]
        if grad > gradmax:
            gradmax = grad
        
        if np.allclose([y_2d(x_smooth[:-2])[i]], [0.], atol=0.001) and y_func(x_smooth)[i] > 0.1 * np.amax(tauc_spectrum[:, 1]) and grad >= gradmax:
            x_0 = x_smooth[i]
            y_0 = y_func(x_smooth)[i]

    # Calculate extrapolation line (y = mx + c)
    m = y_1d(x_0)
    c = y_0 - m * x_0

    # Calculate optical gap (x-intercept, where y = 0)
    x_cross = -c / m

    # --- ERROR ANALYSIS ---
    # Perform linear regression on the fitted data for error estimation
    from scipy.stats import linregress
    
    # Use a reasonable window around the inflection point for regression
    window_size = int(len(x_smooth) * 0.3)  # 30% of data points
    idx_window = np.abs(x_smooth - x_0) < (x_smooth[-1] - x_smooth[0]) * 0.15
    
    if np.sum(idx_window) > 2:
        x_window = x_smooth[idx_window]
        y_window = y_func(x_window)
        
        # Linear regression for error calculation
        slope, intercept, r_value, p_value, std_err = linregress(x_window, y_window)
        
        # Calculate residuals and standard error of the estimate
        y_fitted = slope * x_window + intercept
        residuals = y_window - y_fitted
        n = len(residuals)
        se_estimate = np.sqrt(np.sum(residuals**2) / max(n - 2, 1))
        
        # Error in slope and intercept using standard formulas
        x_mean = np.mean(x_window)
        sxx = np.sum((x_window - x_mean)**2)
        se_slope = se_estimate / np.sqrt(sxx) if sxx > 0 else 0
        se_intercept = se_estimate * np.sqrt(1/n + x_mean**2 / sxx) if sxx > 0 else 0
        
        # Error propagation: x_cross = -c/m, where c = intercept, m = slope
        # Using: σ_x = sqrt((∂x/∂c * σ_c)^2 + (∂x/∂m * σ_m)^2)
        # ∂x/∂c = -1/m, ∂x/∂m = c/m^2
        if abs(slope) > 1e-10:
            dxdc = -1 / slope
            dxdm = intercept / (slope**2)
            x_cross_error = np.sqrt((dxdc * se_intercept)**2 + (dxdm * se_slope)**2)
        else:
            x_cross_error = np.inf
    else:
        x_cross_error = np.nan
        slope = m
        intercept = c

    # Plotting
    ax.set_xlabel(r'Energy (eV)')
    ax.set_ylabel(r'$(\ln(1-EQE) \cdot E)^2 \ (eV)^2$')
    # ax.text(0.05, 0.9, f'Band gap $E_g$ = {x_cross:.4f} ± {x_cross_error:.4f} eV', transform=ax.transAxes, fontsize=11, fontweight='bold')
    ax.set_title('Linear fitting EQE from ' + title)

    if df is not None:
        # Store the result and error in the DataFrame using positional indexing
        df.iloc[0, df.columns.get_loc('Eg (eV) - linear')] = x_cross
        # Add error column if it exists
        if 'Error (eV) - linear' in df.columns:
            df.iloc[0, df.columns.get_loc('Error (eV) - linear')] = x_cross_error

    # Define coordinates for the tangent line for plotting
    y_max_plot = 0.7 * np.amax(tauc_spectrum[:, 1])
    x_line_start = -c / m
    x_line_end = (y_max_plot - c) / m

    ax.plot(x_tot, y_tot, '-', label='Processed Data')
    ax.plot([x_line_start, x_line_end], [0, y_max_plot], '--', label='Tangent Fit')
    ax.plot(tauc_spectrum[:, 0], np.zeros(len(tauc_spectrum)), '-', color='gray')
    ax.plot(x_cross, 0, 'ro', label=f'Optical Gap: {x_cross:.3f} ± {x_cross_error:.3f} eV')
    
    ax.legend()
    
    save_name = 'EQE/Eg_from_linear_fitting_' + title.replace('/', '-').replace('*', 'x') + '.png'
    plt.savefig(save_name, bbox_inches='tight')
    plt.show()
    
    return x_cross, x_cross_error

# --- Main Execution ---
# Band-gap extraction from EQE data
low_limit = 850

# Ensure the EQE directory exists for saving files
Path('EQE').mkdir(exist_ok=True, parents=True)

# Output DataFrame setup with proper index - now includes error column
df = pd.DataFrame(columns=['Eg (eV) - linear', 'Error (eV) - linear'], index=[0])
title = 'Sample_Data' # You can change this to whatever you want the sample to be called

try:
    # Use the already-loaded eqe_df DataFrame from previous cell
    data = eqe_df.copy()
    
    # Filter raw data by wavelength limit
    xdata_raw = data['WL'][data['WL'] >= low_limit].values
    ydata_raw = data['QE'][data['WL'] >= low_limit].values
    
    # Prevent log(0) or negative logs if QE equals or exceeds 100%
    ydata_clipped = np.clip(ydata_raw, 0, 99.999)
    
    # Convert Wavelength (nm) to Energy (eV)
    energy_ev = 1239.84193 / xdata_raw
    
    # Calculate logarithmic relation: (ln(1 - EQE/100) * E)^2
    y_metric = (np.log(1 - (ydata_clipped / 100)) * energy_ev)**2
    
    # Perform fitting and plotting
    eg_value, eg_error = linear_fitting(energy_ev, y_metric, title, df)
    print(f"✓ Band-gap extraction completed successfully")
    print(f"  Eg = {eg_value:.4f} ± {eg_error:.4f} eV")

except Exception as error:
    print(f'ERROR WITH THE LINEAR FITTING:')
    print(error)
    import traceback
    traceback.print_exc()
```
